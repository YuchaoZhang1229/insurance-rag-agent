### 第 1 步：程序入口最早处调用 `configure_logging`

```python
# main.py
from app.core.logging import configure_logging, get_logger
from app.core.config import settings   # 你前面那个 pydantic Settings

configure_logging(level=settings.logging.level)   # 比如 "INFO" / "DEBUG"

logger = get_logger(__name__)
logger.info("应用启动完成")
```

### 第 2 步：在任何业务模块里 `get_logger` 取日志器写日志

```python
from app.core.logging import get_logger

logger = get_logger(__name__)   # __name__ 会作为 logger 名出现在输出里

logger.info("用户发起咨询", product="百万医疗", amount=1000000)
```

### 一句话记忆

- 启动时 `configure_logging(settings.log_level)` 配一次；
- 之后每个模块 `get_logger(__name__)` 取一个；
- 普通日志直接 `logger.info(...)`，要带字段就传 `kwargs`
